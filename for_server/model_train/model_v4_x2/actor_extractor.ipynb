```python
import torch
from sb3_contrib import RecurrentPPO
```

```python
model = RecurrentPPO.load("variable_leverage_model_v4_phase_2/ppo_model_phase_2", device="cpu")

policy = model.policy

print(policy)

actor_state = {
    "features_extractor": policy.pi_features_extractor.state_dict(),
    "lstm_actor": policy.lstm_actor.state_dict(),
    "policy_net": policy.mlp_extractor.policy_net.state_dict(),
    "action_net": policy.action_net.state_dict(),
}
display(actor_state)
```

```python
b = policy.lstm_actor.bias_hh_l0 + policy.lstm_actor.bias_ih_l0
b_i, b_f, b_g, b_o = torch.chunk(b, 4)
print(b_f.mean())
```

```python
b = policy.lstm_critic.bias_hh_l0 + policy.lstm_critic.bias_ih_l0
b_i, b_f, b_g, b_o = torch.chunk(b, 4)
print(b_f.mean())
```
