```python
import torch
from sb3_contrib import RecurrentPPO
```

```python
model = RecurrentPPO.load("variable_leverage_model_v3_phase_2/ppo_model_phase_2", device="cpu")

policy = model.policy

print(policy)

actor_state = {
    "features_extractor": policy.features_extractor.state_dict(),
    "lstm_actor": policy.lstm_actor.state_dict(),
    "policy_net": policy.mlp_extractor.policy_net.state_dict(),
    "action_net": policy.action_net.state_dict(),
}
display(actor_state)
```

```python
torch.save(actor_state, "actor_weights_v2.pth")
```
