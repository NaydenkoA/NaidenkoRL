```python
import numpy as np
from leverage_sim import simulate_single_episode_v2
from utils import plot
```

```python
def simulate_several_episodes(sigma):
    n = 500
    s = 0

    for i in range(n):
        s = s + simulate_single_episode_v2(sigma)
    
    return s/n
```

```python
def collect_statistic(sigma0, sigma1, n):
    delta = (sigma1 - sigma0)/n

    sigmas = []
    intervantions = []

    for i in range(n):
        sigma = sigma0 + delta*i
        intervantion = simulate_several_episodes(sigma)

        sigmas.append(sigma)
        intervantions.append(intervantion)

    return sigmas, intervantions
```

```python
sigma, interventions = collect_statistic(0.01, 0.5, 100)

plot(
    series = [interventions],
    times = sigma,
    x = 'sigma',
    y = 'bot intervation, %'
)
```
