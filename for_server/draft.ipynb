```python
import math
```

```python
def get_looped_leverage(q,n):
    return (1-q**n)/(1-q)
```

```python
def get_loop_naive(q, min_leverage):
    n = int(math.log(1-min_leverage*(1-q))/math.log(q))
    return n
```

```python
def get_prev_loop(q, current_looped_leverage):
    return 1/q*(current_looped_leverage - 1)

def get_next_loop(q, current_looped_leverage):
    return q*current_looped_leverage + 1
```

```python
def get_boundary_leverage(current_min_leverage, min_cr, current_calm_part):
    y = current_min_leverage - 1
    x = (1 - y*(min_cr - 1))/min_cr

    return 1 + y + x/current_calm_part
```

```python
def get_loop(q, min_leverage, min_cr, current_calm_part):
    n = get_loop_naive(q, min_leverage)
    current_lev = get_looped_leverage(q, n)
    boundary_leverage = get_boundary_leverage(current_lev, min_cr, current_calm_part)
    if boundary_leverage > min_leverage:
        while True:
            new_current_lev = get_prev_loop(q, current_lev)
            new_n = n - 1
            new_boundary_leverage = get_boundary_leverage(new_current_lev, min_cr, current_calm_part)
            if new_boundary_leverage < min_leverage:
                break
            else:
                current_lev = new_current_lev
                n = new_n
                boundary_leverage = new_boundary_leverage
                if n == 1:
                    break
    return n, get_looped_leverage(q, n)
```
